### Create a basic random forest model on the training set.

```python
rf_basic = RFC(max_features=len(indep_vars),
               criterion='log_loss',
               n_estimators=50,
               max_depth=10,
               min_samples_leaf=1,
               bootstrap=True,
               oob_score=True,
               random_state=seed)
rf_basic.fit(X_train, y_train)
```

### We'll use this function to quickly calculate the log loss of a model.

```python
def log_loss(y_true, y_prob, eps=1e-15):
    y_true = np.array(y_true)
    y_prob = np.array(y_prob)
    
    # avoid log(0)
    y_prob = np.clip(y_prob, eps, 1 - eps)
    
    ll = -np.mean(y_true * np.log(y_prob) + (1 - y_true) * np.log(1 - y_prob))
    return ll
```

### Compute the log loss on the training set.

```python
y_hat_basic_train = rf_basic.predict_proba(X_train)[:,1]
print('train log loss:', log_loss(y_train, y_hat_basic_train))
```

### Calculate the out-of-bag error (bootstrapping) on training data to estimate the log loss on the test set.

```python
y_hat_basic_oob = rf_basic.oob_decision_function_[:,1]
print('oob log loss:  ', log_loss(y_train, y_hat_basic_oob))
```

### Compute the log loss on the test set.

```python
y_hat_basic_test = rf_basic.predict_proba(X_test)[:,1]
print('test log loss: ', log_loss(y_test, y_hat_basic_test))
```

### Build an improved random forest model on this same set of data. 

You will be scored according to how well it performs on an unseen holdout set.

```python
# Documentation: https://scikit-learn.org/dev/modules/generated/sklearn.ensemble.RandomForestClassifier.html

# Choose hyperparameters by out-of-bag log loss (training data only; the test set is never used for tuning)
best_oob, best_params = np.inf, None
for max_depth in [3, 5, 8]:
    for min_samples_leaf in [20, 50, 100]:
        for max_features in ['sqrt', 0.5]:
            m = RFC(n_estimators=300,
                    criterion='log_loss',
                    max_depth=max_depth,
                    min_samples_leaf=min_samples_leaf,
                    max_features=max_features,
                    bootstrap=True,
                    oob_score=True,
                    random_state=seed,
                    n_jobs=-1).fit(X_train, y_train)
            score = log_loss(y_train, m.oob_decision_function_[:,1])
            if score < best_oob:
                best_oob, best_params = score, dict(max_depth=max_depth,
                                                    min_samples_leaf=min_samples_leaf,
                                                    max_features=max_features)
print(best_params, best_oob)

rf_better = RFC(n_estimators=1000,
                criterion='log_loss',
                bootstrap=True,
                oob_score=True,
                random_state=seed,
                n_jobs=-1,
                **best_params)
rf_better.fit(X_train, y_train)
```

### Calculate the out-of-bag log loss for this model. Store it as `log_loss_oob`

```python
y_hat_better_oob = rf_better.oob_decision_function_[:,1]

log_loss_oob = log_loss(y_train, y_hat_better_oob)
print(log_loss_oob)
```

### Create the feature importance list for this model.

```python
feature_imp = pd.DataFrame(
    {'importance':rf_better.feature_importances_},
    index=indep_vars)
feature_imp
```

## Discussion Questions

### (1) Explain what "value" means in the decision tree plots.

It shows how many training records in that node fall into each class: [no heart attack risk, heart attack risk]. The tree predicts whichever class has more records in the node.

### (2) How is “log loss” different from the log likelihood we defined in class?

Log likelihood is a score we try to make as large as possible when fitting a model. Log loss is the negative of the average log likelihood, so it is an error we want as small as possible. We use it to compare how well models' predicted probabilities match the actual outcomes.

### (3) Explain how we know the basic model is overfit before attempting to build any other models.

The training log loss (0.47) is much lower than the out-of-bag log loss (0.67). The model does great on data it was trained on but much worse on data its trees did not see, so it is memorizing the training data instead of learning general patterns.
